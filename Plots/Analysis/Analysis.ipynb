```python
import pandas as pd
import matplotlib.pyplot as plt

# Nombre del archivo CSV
nombre_archivo_csv = r"C:\Users\joc_h\lightcdmx\Plots\Analysis\Multiplot.csv"

# Columnas del CSV
columna_x = 'columna_x'  # Eje X común
columna_y1 = 'columna_y1'  # Primer Eje Y
columna_y2 = 'columna_y2'  # Segundo Eje Y
columna_y3 = 'columna_y3'  # Tercer Eje Y
columna_y4 = 'columna_y4'  # Cuarto Eje Y

# Etiquetas para los ejes Y
etiqueta_y1 = 'Unidades 1'
etiqueta_y2 = 'Unidades 2'
etiqueta_y3 = 'Unidades 3'
etiqueta_y4 = 'Unidades 4'

try:
    # Leer el archivo CSV
    data = pd.read_csv(nombre_archivo_csv)

    # Obtener los datos
    x_data = data[columna_x]
    y1_data = data[columna_y1]
    y2_data = data[columna_y2]
    y3_data = data[columna_y3]
    y4_data = data[columna_y4]

    # Crear la figura y el primer conjunto de ejes
    fig, ax1 = plt.subplots(figsize=(12, 8))  # Ajusta el tamaño de la figura si es necesario

    # Graficar la primera serie de datos (Eje Y principal)
    color1 = 'tab:blue'
    ax1.set_xlabel('Eje X')
    ax1.set_ylabel(etiqueta_y1, color= '#808080')
    ax1.plot(x_data, y1_data, marker='o', linestyle='-', color='#808080', linewidth=2, label='Serie 1')
    ax1.tick_params(axis='y', labelcolor= '#808080')
    ax1.grid(False)

    # Crear el segundo conjunto de ejes (Eje Y secundario 1)
    ax2 = ax1.twinx()
    ax2.set_ylabel(etiqueta_y2, color='black')
    ax2.plot(x_data, y2_data, marker='s', linestyle='--', color='black', linewidth=2, label='Serie 2')
    ax2.tick_params(axis='y', labelcolor='black')

    # Crear el tercer conjunto de ejes (Eje Y secundario 2)
    ax3 = ax1.twinx()
    # Mover el eje del tercer conjunto a la derecha
    ax3.spines['right'].set_position(('outward', 60))  # Ajusta la posición según sea necesario
    ax3.set_ylabel(etiqueta_y3, color="red")
    ax3.plot(x_data, y3_data, marker='^', linestyle='-.', color="red", linewidth=2, label='Serie 3')
    ax3.tick_params(axis='y', labelcolor="red")

    # Crear el cuarto conjunto de ejes (Eje Y secundario 3)
    ax4 = ax1.twinx()
    # Mover el eje del cuarto conjunto a la derecha
    ax4.spines['right'].set_position(('outward', 120))  # Ajusta la posición según sea necesario
    color4 = 'tab:blue'
    ax4.set_ylabel(etiqueta_y4, color=color4)
    ax4.plot(x_data, y4_data, marker='D', linestyle=':', color=color4, linewidth=2, label='Serie 4')
    ax4.tick_params(axis='y', labelcolor=color4)

    # Asegurar que los ejes estén en el frente
    for ax in [ax1, ax2, ax3, ax4]:
        ax.set_zorder(10)
        ax.patch.set_visible(False)
        

    # Añadir título
    plt.title('Gráfica con Cuatro Ejes Y')
    
    # Guardar la figura
    plt.savefig('grafica_cuatro_ejes.png', dpi=400)

    # Mostrar la gráfica
    plt.show()

except FileNotFoundError:
    print(f"Error: El archivo '{nombre_archivo_csv}' no fue encontrado.")
except KeyError as e:
    print(f"Error: La columna '{e}' no existe en el archivo CSV.")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```
